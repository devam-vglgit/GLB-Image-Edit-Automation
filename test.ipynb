```python
import cv2
import numpy as np
```

```python


def recenter_on_white(img_path, out_path, canvas_size=1024, margin_ratio=0.05):
    img = cv2.imread(img_path)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Anything not near-white is "object"
    _, mask = cv2.threshold(gray, 245, 255, cv2.THRESH_BINARY_INV)
    coords = cv2.findNonZero(mask)
    x, y, w, h = cv2.boundingRect(coords)

    # Crop tightly to the object
    obj = img[y:y+h, x:x+w]

    # Scale object to fit canvas with margin
    scale = (canvas_size * (1 - margin_ratio)) / max(w, h)
    new_w, new_h = int(w * scale), int(h * scale)
    obj_resized = cv2.resize(obj, (new_w, new_h))

    # Paste centered on white canvas
    canvas = np.full((canvas_size, canvas_size, 3), 255, dtype=np.uint8)
    x_off = (canvas_size - new_w) // 2
    y_off = (canvas_size - new_h) // 2
    canvas[y_off:y_off+new_h, x_off:x_off+new_w] = obj_resized

    cv2.imwrite(out_path, canvas)
```

```python
recenter_on_white('gem-ai-01-8683978.png', 'output_image.png')
```
